```python
from langchain_text_splitters import (
    MarkdownTextSplitter,
    PythonCodeTextSplitter
)
```

```python
document = """
# Virat Kohli: The Modern Cricket Icon

**Early Life**
Virat Kohli, born on November 5, 1988, in Delhi, India, is one of the most celebrated cricketers in the modern era. 
His father, Prem Kohli, enrolled him in the West Delhi Cricket Academy at age nine.

**Rise to Stardom**
He led the Indian U-19 team to World Cup victory in 2008, making his international debut later that year.

**Peak Years**
Between 2016 and 2019, Kohli dominated world cricket, breaking numerous records. 
He became the fastest player to score 8,000–11,000 ODI runs.

**Leadership**
Under his captaincy, India won its first Test series in Australia (2018–19) and topped ICC Test rankings.

**Off the Field**
He runs the Virat Kohli Foundation supporting underprivileged children and sports in India.

**Legacy**
Even after stepping down as captain in 2022, he remains one of the most influential cricketers worldwide.
"""
```

```python
md_splitter = MarkdownTextSplitter(chunk_size=200,chunk_overlap=0)
md_chunks = md_splitter.split_text(document)
print("\n MarkDownTextSplitter Output")
for i,c in enumerate(md_chunks,1):
    print(f"chunk {i}:\n{c}\n{'-'*60}")
```

### Code-based Chunking

```python
code_doc = f'''
"""
{document}
"""
def summary():
    return "Virat Kohli - modern cricket icon with leadership and batting excellence."
'''
```

```python
py_splitter = PythonCodeTextSplitter(chunk_size=200,chunk_overlap=0)
py_chunks = py_splitter.split_text(code_doc)
print("\n PythonCodeTextSplitter Output ")
for i,c in enumerate(py_chunks,1):
    print(f"chunk {i}\n{c}\n{'-'*60}")
```
